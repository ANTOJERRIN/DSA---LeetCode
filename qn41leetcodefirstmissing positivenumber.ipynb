```python
# the c++ ccode implementation is 
class Solution {
public:
    int firstMissingPositive(vector<int>& nums) {
        int n = nums.size();
        
        for (int i = 0; i < n; i++) {
            // While the current number is in range [1, n] 
            // AND it's not already sitting at its correct index
            while (nums[i] > 0 && nums[i] <= n && nums[nums[i] - 1] != nums[i]) {
                swap(nums[i], nums[nums[i] - 1]);
            }
        }
        
        // Scan to find the first misplaced number
        for (int i = 0; i < n; i++) {
            if (nums[i] != i + 1) {
                return i + 1;
            }
        }
        
        // If the array was perfectly [1, 2, 3, ..., n]
        return n + 1;
    }
};
```
