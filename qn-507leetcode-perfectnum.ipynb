```python
# the python code soln:
class Solution:
    def checkPerfectNumber(self, num: int) -> bool:
        if(num<=1):
            return False
        s=1
        for i in range(2,int(num**0.5)+1): #the thing is here range fun ends with n-1 not with just n 
            if(num%i==0):
                s=s+i
                if((num//i)!=i):
                    s=s+(num//i)
        return s==num            
```

```python
# the c++ code soln is:
class Solution {
public:
    bool checkPerfectNumber(int num) {
        if(num<=1) return false;
        int s=1;
        list<int> ls;
        for(int i=2;i<=sqrt(num);i++)
        {
            if(num%i==0)
            {
               
                ls.push_back(i);
                s+=i;
                if((num/i)!=i)
                {
                ls.push_back(num/i);
                s+=(num/i);
                }
            }
           
        }
        return s==num;
        
    }
};
```
