```python
# the c++ code is :
class Solution {
public:
    string convertToTitle(int columnNumber) {
        string r="";
        while(columnNumber>0)
        {
            columnNumber--;
            r=r+(char)('A'+(columnNumber %26));
            columnNumber=columnNumber/26;
        }
        reverse(r.begin(),r.end());

        return r;
        
    }
};
```

```python
# the python code is:
class Solution:
    def convertToTitle(self, columnNumber: int) -> str:
        r = ""
        while columnNumber > 0:
            columnNumber -= 1
            
            # 1. Get the current character using ord() and chr()
            current_char = chr(ord('A') + (columnNumber % 26))
            
            # 2. Add it to the FRONT of the string
            r = current_char + r
            
            # 3. Integer divide by 26 for the next iteration
            columnNumber //= 26
            
        return r
```
