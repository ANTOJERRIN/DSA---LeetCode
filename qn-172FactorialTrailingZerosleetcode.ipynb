```python
# the python code is :
class Solution:
    def trailingZeroes(self, n: int) -> int:
        c=0
        while(n>0):
            n=n//5
            c=c+n
        return c
```

```python
# the c++ solution
class Solution {
public:
    int trailingZeroes(int n) {
        int c=0;
        while(n>0)
        {
            n/=5;
            c+=n;
        }
        return c;
        
        
    }
};   
```
