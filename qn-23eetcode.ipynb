```python
import heapq

class Solution:
    def mergeKLists(self, lists: List[Optional[ListNode]]) -> Optional[ListNode]:
        heap = []
        
        # 1. Initialize the heap with the head of each linked list.
        for i, lst in enumerate(lists):
            if lst:
                # Push a tuple: (value, tie-breaker index, node)
                heapq.heappush(heap, (lst.val, i, lst))
                
        dummy = ListNode(0)
        curr = dummy
        
        # 2. Extract the smallest node and push the next node from its liṇṇst.
        while heap:
            val, i, node = heapq.heappop(heap)
            curr.next = node
            curr = curr.next
            
            # If the popped node has a next node, push it into the heap
            if node.next:
                heapq.heappush(heap, (node.next.val, i, node.next))
                
        return dummy.next
```
