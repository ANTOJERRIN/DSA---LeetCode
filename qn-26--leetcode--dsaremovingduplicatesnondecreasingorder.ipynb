```python
# the c++ code implementation
class Solution {
public:
    int removeDuplicates(vector<int>& nums) {
        if(nums.empty()) return 0;//validation check
        int k=1;//the first pointer 
        for(int i=1;i<nums.size();i++)
        {
            if(nums[i]!=nums[i-1])//second pointer
            {
                nums[k]=nums[i];
                k++;//incrementing the first pointer
            }
        }
        return k;//returning the no of unique elements and the op in array format
        
    }
};
```

```python
# the python code implementation is :
class Solution:
    def removeDuplicates(self, nums: list[int]) -> int:
        n=len(nums) #taking the length
        if(n<0): # validation check
            return 0
        k=1 #the first pointer for overwriting
        for i in range(1,n,1): #second pointer for reading the unique elements 
            if(nums[i]!=nums[i-1]):
                nums[k]=nums[i] #overwriting the duplicate elements with unique elements getting from second pointer
                k=k+1
        return k            
```
