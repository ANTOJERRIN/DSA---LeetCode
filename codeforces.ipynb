```python
def solve_euclid_sequence(n, b):
    if n == 2:
        return max(b[0], b[1]), min(b[0], b[1])
    
    # For n >= 3, sequence must be strictly decreasing
    b_sorted = sorted(b, reverse=True)
    
    for i in range(n - 2):
        if b_sorted[i] <= b_sorted[i+1] or b_sorted[i+1] <= b_sorted[i+2]:
            return -1
        if b_sorted[i] % b_sorted[i+1] != b_sorted[i+2]:
            return -1
            
    return b_sorted[0], b_sorted[1]

# Test with example inputs
test_cases = [
    (2, [1, 1]),
    (2, [1, 2]),
    (4, [1, 2, 3, 4]),
    (3, [6, 4, 2]),
    (4, [3, 8, 13, 5]),
    (3, [1, 1, 1])
]

for n, b in test_cases:
    res = solve_euclid_sequence(n, b)
    if res == -1:
        print(-1)
    else:
        print(f"{res[0]} {res[1]}")
```
