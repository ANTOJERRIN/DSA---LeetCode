```python
# the c++ code implementation is 
class Solution {
public:
    bool areOccurrencesEqual(string s) {
        int n=s.size();
        unordered_map<char,int> h;
        for(char t:s)
        {
            h[t]++;

        }
        int f=h[s[0]]; //intially take the count of the freq as starting character so that we can compare with each element in the array to get the bool
        for(char t:s)
        {
            if(h[t]!=f) return false;
        }
        return true;
    }
};
```

```python
# the python code implementation  is 
class Solution:
    def areOccurrencesEqual(self, s: str) -> bool:
        n = len(s)
        u = {}

        for i in range(n):
            u[s[i]] = u.get(s[i], 0) + 1

        f = u[s[0]]

        for i in range(n):
            if u[s[i]] != f:
                return False

        return True
```
