```python
class Solution:
    def sumGame(self, num: str) -> bool:
        n=len(num)
        q,s=0,0.0
        for i in range(n//2):
            if(num[i]=='?'):
                q=q+1
            else:
                s+=int(num[i])
        for i in range(n//2,n):
            if(num[i]=='?'):
                q=q-1
            else:
                s-=int(num[i])
        return s+(q/2)*9!=0                            
        
```

```python
class Solution {
public:
    bool sumGame(string num) {
        int len=num.length();
        double sumdiff=0;
        int qdiff=0;
        for(int i=0;i<len/2;i++)
        {
            if(num[i]=='?')
            {
                qdiff++;
            }
            else{
                sumdiff+=(num[i]-'0');
            }
        }
        for(int i=len/2;i<len;i++)
        {
            if(num[i]=='?')
            {
                qdiff--;
            }
            else
            {
                sumdiff-=(num[i]-'0');
            }

        }
        return sumdiff+(qdiff/2.0)*9!=0;
    }
};
```
