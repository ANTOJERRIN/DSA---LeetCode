```python
# the python code implementation:
class Solution:
    def sumCounts(self, nums: List[int]) -> int:
        n=len(nums)
        t=0
        for i in range(0,n):
            s=[]
            for j in range(i,n):
                s.append(nums[j])
                f=len(set(s))
                t=t+f*f
        return t        
        
```

```python
# the c++ code implementation is :
class Solution {
public:
    int sumCounts(vector<int>& nums) {
        int n=nums.size(); int t=0;//first we are taking here the size
        for(int i=0;i<n;i++)
        {
            unordered_set<int> u;//hash set 
            for(int j=i;j<n;j++)
            {
                u.insert(nums[j]);
                int  d=u.size();
                t+=d*d;
                
            }
          
        }
        return t;
        
    }
};
```
