```python
# c++ code : dsa
class Solution {
public:
    bool isRectangleOverlap(vector<int>& rec1, vector<int>& rec2) {
        bool v=max(rec1[0],rec2[0])<min(rec1[2],rec2[2]);
        bool v1=max(rec1[1],rec2[1])<min(rec1[3],rec2[3]);

        return v&&v1;
        
    }
};
```

```python
# python code 
class Solution:
    def isRectangleOverlap(self, rec1: List[int], rec2: List[int]) -> bool:
        v=(max(rec1[0],rec2[0]))<(min(rec1[2],rec2[2]))
        v1=(max(rec1[1],rec2[1]))<(min(rec1[3],rec2[3]))

        return v and v1
        
```
