```python
# this is the python soluntion
class Solution:
    def missingMultiple(self, nums: List[int], k: int) -> int:
        l=len(nums)
        m=1
        while(1):
            t=m*k
            f=False

            for i in range(l):
                if nums[i]==t:
                    f=True
                    break
            if(not f):
                return t
            m=m+1    
```

```python
# this tis the fundamentl c+++ soluntion
class Solution {
public:
    int missingMultiple(vector<int>& nums, int k) {
        int m=1,t;bool f;
        while(true)
        {
            t=m*k;
            f=false;
        
        for(int i=0;i<nums.size();i++)
        {
            if(nums[i]==t)
            {
                f=true;
                break;
            }
        }
        if(!f)
        {
            return t;
        }
        m++;
        }
    }
};
```
