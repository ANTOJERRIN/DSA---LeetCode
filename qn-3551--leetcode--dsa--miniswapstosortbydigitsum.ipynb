```python
# the c++ code implementation is :
#include <vector>
#include <algorithm>

class Solution {
public:
    int minSwaps(std::vector<int>& nums) {
        int n = nums.size();
        
        // 1. Calculate digit sums without mutating original values
        std::vector<std::pair<int, int>> sortedNums(n); // {digit_sum, value}
        for (int i = 0; i < n; i++) {
            int val = nums[i];
            int c = 0;
            while (val > 0) {
                int d = val % 10;
                c += d;
                val = val / 10;
            }
            sortedNums[i] = {c, nums[i]};
        }

        // 2. Pair elements with their initial indices before sorting
        std::vector<std::pair<std::pair<int, int>, int>> indexed(n);
        for (int i = 0; i < n; i++) {
            indexed[i] = {sortedNums[i], i};
        }

        // 3. Sort by custom rules (digit sum first, then value)
        std::sort(indexed.begin(), indexed.end());

        // 4. Count permutation cycles to find min swaps
        std::vector<bool> visited(n, false);
        int swaps = 0;

        for (int i = 0; i < n; i++) {
            if (visited[i] || indexed[i].second == i) {
                continue;
            }

            int cycleSize = 0;
            int curr = i;
            while (!visited[curr]) {
                visited[curr] = true;
                curr = indexed[curr].second;
                cycleSize++;
            }

            if (cycleSize > 0) {
                swaps += (cycleSize - 1);
            }
        }

        return swaps;
    }
};
```
