```python
# the c++ code implementation:
class Solution {
private:
    bool isVowel(char c) {
        return c == 'a' || c == 'e' || c == 'i' || c == 'o' || c == 'u' ||
               c == 'A' || c == 'E' || c == 'I' || c == 'O' || c == 'U';
    }

public:
    string reverseVowels(string s) {
        int left = 0;
        int right = s.size() - 1;

        while (left < right) {
            // Advance left pointer until it points to a vowel
            while (left < right && !isVowel(s[left])) {
                left++;
            }
            // Retract right pointer until it points to a vowel
            while (left < right && !isVowel(s[right])) {
                right--;
            }
            // Swap the identified vowels
            if (left < right) {
                swap(s[left], s[right]);
                left++;
                right--;
            }
        }

        return s;
    }
};
# this needs to be worked on
```
