```python
# the python code implementation is:
class Solution:
    def moveZeroes(self, nums: list[int]) -> None:
        """
        Do not return anything, modify nums in-place instead.
        """
        n=len(nums)
        j=0 #the first pointer
        t=0
        for i in range(0,n): #the ith var is the second pointer
            if nums[i]!=0:
                t=nums[j]
                nums[j]=nums[i]
                nums[i]=t
                j=j+1
        
```

```python
# the c++ code implementation is :
class Solution {
public:
    void moveZeroes(vector<int>& nums) {
        int n=nums.size();
        int l=0;
        for(int r=0;r<n;r++)
        {
            if(nums[r]!=0)
            {
                swap(nums[l],nums[r]);
                l++;
            }
        }
        
    }
};
```
