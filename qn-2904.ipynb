```python
# c++ code
class Solution {
public:
    string shortestBeautifulSubstring(string s, int k) {
        int n = s.length();
        string best_ans = "";
        int min_len = n + 1; // Initialize with a value larger than max possible length

        for (int i = 0; i < n; i++) {
            int count_ones = 0;
            string current_sub = "";
            
            for (int j = i; j < n; j++) {
                current_sub += s[j];
                if (s[j] == '1') {
                    count_ones++;
                }
                
                // If we found a beautiful substring
                if (count_ones == k) {
                    int curr_len = current_sub.length();
                    
                    // Rule 1: Found a strictly shorter beautiful substring
                    if (curr_len < min_len) {
                        min_len = curr_len;
                        best_ans = current_sub;
                    }
                    // Rule 2: Same length, but check if it's lexicographically smaller
                    else if (curr_len == min_len) {
                        if (current_sub < best_ans) {
                            best_ans = current_sub;
                        }
                    }
                    break; // No need to expand further from this starting point `i`
                }
            }
        }
        return best_ans;
    }
};
```
