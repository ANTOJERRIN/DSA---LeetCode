```python
# the c++ code implementation is :
class Solution {
public:
    vector<int> intersection(vector<int>& nums1, vector<int>& nums2) {
       unordered_set<int> set1(nums1.begin(),nums1.end());  //here we re builting  a set for counter fucntion
       vector<int> r(0);//the result var ini to 0
       for(auto i:nums2)//looping from the nums2 array
       {
        if(set1.count(i))//if the element is present in the nums1 tht is the count is more means then
        {
            r.push_back(i);//meqan wwe repushing the element in the array
            set1.erase(i);//adn we have to erase the eement which is alreasy present in the set 1
        }
    
       }
       return r; 
    }
};
```

```python
# the python code implementation is:
class Solution:
    def intersection(self, nums1: list[int], nums2: list[int]) -> list[int]:
        r=[]
        for i in nums2:
            if(nums1.count(i) and i not in r):
                r.append(i)
                nums1.remove(i)
        return r                  
        
```
