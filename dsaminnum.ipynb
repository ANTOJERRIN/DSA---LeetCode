```python
class Solution(object):
    def minimumPushes(self, word):
        """
        :type word: str
        :rtype: int
        """
        n = len(word)
        pushes = 0
        
        # 1st push group: up to 8 letters
        if n > 0:
            pushes += min(n, 8) * 1
        
        # 2nd push group: letters 9 to 16
        if n > 8:
            pushes += min(n - 8, 8) * 2
            
        # 3rd push group: letters 17 to 24
        if n > 16:
            pushes += min(n - 16, 8) * 3
            
        # 4th push group: letters 25 to 26
        if n > 24:
            pushes += (n - 24) * 4
            
        return pushes
```
