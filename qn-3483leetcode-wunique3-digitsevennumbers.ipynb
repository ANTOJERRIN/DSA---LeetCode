```python
# this is the c++ code implementation:
class Solution {
public:
    int totalNumbers(vector<int>& digits) {
        vector<int> count(10,0);
        for(auto it:digits)
        {
            count[it]++;
            
        }
        int v=0;
        for(int i=100;i<=998;i+=2)
        {
            int d1=i/100;
            int d2=(i/10)%10;
            int d3=i%10;
            vector<int> temp(10,0);
            temp[d1]++;
            temp[d2]++;
            temp[d3]++;

            if(temp[d1] <= count[d1] && temp[d2]<=count[d2] && temp[d3]<=count[d3])
            {
                v++;
            }
        }
        return v;
        
    }
};
```

```python
# this is the pyhton code implementation
class Solution:
    def totalNumbers(self, digits: List[int]) -> int:
        c=[0]*10
        for i in digits:
            c[i]=c[i]+1
        v=0
        for j in range(100,1000,2):
            t=[0]*10
            d1=j//100
            d2=(j//10)%10
            d3=j%10

            t[d1]=t[d1]+1
            t[d2]=t[d2]+1
            t[d3]=t[d3]+1

            if(t[d1]<=c[d1] and t[d2]<=c[d2] and t[d3]<=c[d3]):
                v=v+1    
        return v
```
