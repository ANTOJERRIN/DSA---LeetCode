```python
# the python code implementation is 
class Solution:
    def pivotIndex(self, nums: list[int]) -> int:
        n=len(nums)
        ls,rs,s=0,0,0
        for i in range(0,n):
            s=s+nums[i]
        for i in range(0,n):
            rs=s-ls-nums[i]
            if(ls==rs):
                return i
            ls=ls+nums[i]        
        return -1    
```

```python
# the c++ code implementation is 
class Solution {
public:
    int pivotIndex(vector<int>& nums) {
        int n=nums.size();
        int ls=0,s=0,rs=0;
        for(int i=0;i<n;i++)
        {
            s+=nums[i];
            
        }
        for(int i=0;i<n;i++)
        {
            rs=s-ls-nums[i];
            if(ls==rs)
            {
                return i;
            }
            ls=ls+nums[i];
        }
        return -1;

        
    }
};
```
