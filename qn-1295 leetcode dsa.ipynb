```python
# the c++ code implementation
class Solution {
public:
    int findNumbers(vector<int>& nums) {
        int n = nums.size();
        int d = 0;
        for(int i = 0; i < n; i++)
        {
            int c = 0; // Reset 'c' to 0 for each new number
            while(nums[i] > 0)
            {
                c++;
                nums[i] = nums[i] / 10;
            }
            if(c % 2 == 0)
            {
                d++;
            }
        }
        return d;
    }
};
```

```python
# the python code implementation
class Solution:
    def findNumbers(self, nums: list[int]) -> int:
        d=0
        n=len(nums)
        for i in range(0,n):
            c=0 #this refreshes for evry var check
            while nums[i]>0:
                c=c+1
                nums[i]=nums[i]//10
            if c%2==0:
                d=d+1    
        return d
```
