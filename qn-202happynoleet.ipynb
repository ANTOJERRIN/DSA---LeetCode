```python
#this is c++ cod ebelow
class Solution {
public:
    bool isHappy(int n) {
        int t = n;
        
        // Happy numbers reach 1, while non-happy numbers always cycle back into 4
        while (t != 1 && t != 4) {
            int s = 0;
            while (t > 0) {
                int r = t % 10;
                s += r * r;
                t = t / 10;
            }
            t = s; // Put the sum back into t for the next round
        }

        return t == 1;
    }
};s
```
