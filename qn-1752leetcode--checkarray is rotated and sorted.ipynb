```python
# the c++ code implementation
class Solution {
public:
    bool check(vector<int>& nums) {
        vector<int> f=nums;
        sort(f.begin(),f.end());
        int n=nums.size();
        for(int i=0;i<n;i++)
        {
            int m=0;
            for(int j=0;j<n;j++)
            {
                if(f[j]==nums[(j+i)%n])
                {
                    m++;
                }
            }
            if(m==n)// size we re checking ///
            {
                return true;
            }
        }
        return false;
        
    }
};
// class Solution {
// public:
//     bool check(vector<int>& nums) {
//         vector<int> f = nums;
//         sort(f.begin(), f.end());
//         int n = nums.size();
        
//         for(int i = 0; i < n; i++) {
//             bool m = true; // Assume it matches for this rotation offset
            
//             for(int j = 0; j < n; j++) {
//                 if(f[j] != nums[(j + i) % n]) {
//                     m = false; // Mismatch found, this offset 'i' is invalid
//                     break;
//                 }
//             }
            
//             if(m) {
//                 return true; // If 'm' is still true, all elements matched
//             }
//         }
//         return false; // Checked all offsets, none worked
//     }
// };
```
