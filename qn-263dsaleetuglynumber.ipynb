```python
class Solution:
    def isUgly(self, n: int) -> bool:
        if(n<=0):
            return False
        t=n
        while(t%2==0):
            t=t/2
        while(t%3==0):
            t=t/3
        while(t%5==0):
            t=t/5

        return t==1                

        
```
