```python
# the pyhton code implementation is 
class Solution:
    def findDisappearedNumbers(self, nums: list[int]) -> list[int]:
        seen=[]
        iu=0
        for i in  range(len(nums)):
            iu=abs(nums[i])-1
            if(nums[iu]>0):
                nums[iu]=-nums[iu]
        for i in range(len(nums)):
            if(nums[i]>0):
                seen.append(i+1)
        return seen        
```

```python
# the c++ code implementation is :
class Solution {
public:
    vector<int> findDisappearedNumbers(vector<int>& nums) {
        int n=nums.size();
        vector<int> v;
        // we re gonnna check wht are the elements prsent in the arr by negating them
        for(int i=0;i<n;i++)
        {
            int in=abs(nums[i])-1;
            if(nums[in]>0)
            {
                nums[in]=-nums[in];
            }

        }
        // finding the missing elemnts present in the array
        for(int i=0;i<n;i++)
        {
            if(nums[i]>0)
            {
                v.push_back(i+1);
            }
        }
        return v;
    
        
    }
};
```
