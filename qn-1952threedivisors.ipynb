```python
# c++ code is:
class Solution {
public:
    bool isThree(int n) {
        int c=0;
        for(int i=1;i<=n;i++)
        {
            if(n%i==0)
            {
                c++;
            }
        }
        return c==3;
        
    }
};
```

```python
# pyhton code is:
class Solution:
    def isThree(self, n: int) -> bool:
        c=0
        for i in range(1,n+1):
            if(n%i==0):
                c=c+1
        return c==3
```
