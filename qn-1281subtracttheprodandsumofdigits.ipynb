```python
# this is the c++ code implementation:
class Solution {
public:
    int subtractProductAndSum(int n) {
        int t=n;
        int p=1,s=0;
        while(t>0)
        {
            int d=t%10;
            p=p*d;
            s=s+d;
            t=t/10;
        }
        return p-s;
        
        
    }
};
```

```python
# this is the python code implementation:
class Solution:
    def subtractProductAndSum(self, n: int) -> int:
        if(n<=0):
            return 0
        t,p,s=n,1,0
        while(t>0):
            d=t%10
            p=p*d
            s=s+d
            t=t//10
        return p-s    
```
