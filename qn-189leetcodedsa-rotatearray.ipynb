```python
# the c++ code implementation
class Solution {
public:
    void rotate(vector<int>& nums, int k) {
        int n=nums.size();
        k=k%n;
        reverse(nums.begin(),nums.end());
        reverse(nums.begin(),nums.begin()+k);
        reverse(nums.begin()+k,nums.end());

        
    }
};
```

```python
# the python code implemetation
class Solution:
    def rotate(self, nums: list[int], k: int) -> None:
        """
        Do not return anything, modify nums in-place instead.
        """
        n=len(nums)
        k=k%n
        nums[:]=nums[-k:]+nums[:-k]
        
    #    remeber here that we re first slicing the list only those three elements from the array backtracking and next slice the array only till tjhe last three elemetns
```
